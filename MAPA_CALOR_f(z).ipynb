```python
import numpy as np
import matplotlib.pyplot as plt

# Definição da função f(z) = 1/(z^2 + a^2)
a = 1  # você pode mudar o valor de 'a' aqui

def f(z):
    return 1 / (z**2 + a**2)

# Criar uma grade de pontos no plano complexo
x = np.linspace(-3, 3, 400)   # eixo real
y = np.linspace(-3, 3, 400)   # eixo imaginário
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y

# Avaliar a função em cada ponto
F = f(Z)

# Evitar explosões no gráfico (na vizinhança dos polos)
F[np.abs(Z - 1j*a) < 1e-2] = np.nan
F[np.abs(Z + 1j*a) < 1e-2] = np.nan

# Plotar mapa de calor do módulo (|f(z)|)
plt.figure(figsize=(6,6))
plt.pcolormesh(X, Y, np.log1p(np.abs(F)), shading='auto', cmap='inferno')
plt.colorbar(label="log(1+|f(z)|)")

# Marcar os polos
plt.plot([0], [a], 'bo', label='z = ia',markersize=14)
plt.plot([0], [-a], 'ro', label='z = -ia', markersize=13)

# Ajustes do gráfico
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")
plt.title("Mapa de calor de $f(z) = 1/(z^2+a^2)$")
plt.legend()
plt.gca().set_aspect("equal")
plt.show()
```
